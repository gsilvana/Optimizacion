```python
# EJERCICIO 4: MODELO LOGÍSTICO DE PESCA
# GINNETH SILVANA MARTÍNEZ URIBE

import numpy as np

H = 80        # Peces pescados por año
K = 1000      # Capacidad del lago
r = 0.4       # Tasa de crecimiento

# Función y significado de cada terminoo
def poblacion_siguiente(P_actual):
    # P_actual       población base
    # r * P_actual   Nacimientos naturales
    # (1 - P/K)      limite ambiental
    # - H            peces pescados

    crecimiento = r * P_actual * (1 - P_actual / K)
    P_sig = P_actual + crecimiento - H
    return P_sig

# Población inicial
P0 = 400
años = 20

# Listas para guardar el historial
historial_P = [P0]

# Simulación  año tras año
for i in range(años):
    # Calculamos la población del siguiente año
    P_sig = poblacion_siguiente(historial_P[i])
    historial_P.append(P_sig)

# Resultado
print (" = " *45)
print ("POBLACIÓN DE PECES EN EL LAGO")
print ("Estado Simulación finalizada")
print ()
print (f"Población Año 0: P0 = {historial_P[0]:.2f} peces")
print (f"Población Año 1: P1 = {historial_P[1]:.2f} peces")
print (f"Población Año 2: P2 = {historial_P[2]:.2f} peces")
print ()
print (f"Población estabilizada en el año {años}: P = {historial_P[-1]:.2f} peces")
print (" = " *45)

# Verificación de los puntos de equilibrio
# Ecuación a resolver: -0.0004*P^2 + 0.4*P - 80 = 0
coeficientes = [-r/K, r, -H]
raices = np.roots(coeficientes)

equilibrio_estable = max(raices)
equilibrio_inestable = min(raices)

print ("Verificación Teórica")
print (f"Equilibrio Estable (Población sana):      {equilibrio_estable:.2f} peces")
print (f"Equilibrio Inestable (Riesgo extinción):  {equilibrio_inestable:.2f} peces")
```
