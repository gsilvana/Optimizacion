```python
# EJERCICIO 1: MEZCLA DE FERTILIZANTES
# GINNETH SILVANA MARTÍNEZ URIBE

# Importaré la librería pulp y utilizaré las clases: maximizar, crear problema y definir variables
from pulp import LpMaximize, LpProblem, LpVariable

# Problema de maximización
maximizar = LpProblem("Fertilizantes", LpMaximize)

# Variables con restriccion de NO negatividad
x1 = LpVariable("F1", lowBound=0, cat='Continuous')
x2 = LpVariable("F2", lowBound=0, cat='Continuous')

# Función objetivo
maximizar += 180*x1 + 220*x2

# Restricciones de disponibilidad
maximizar += 3*x1 + 4*x2 <= 120
maximizar += 2*x1 + x2 <= 80
maximizar += 5*x1 + 3*x2 <= 150

# Resolver
maximizar.solve()

# Resultados
print(" = " *45)
print("GANANCIA MÁXIMA FERTILIZANTES")
print("Estado Optimo encontrado")
print()
print(f"Fertilizante 1  =  {x1.varValue:.2f} toneladas")
print(f"Fertilizante 2  =  {x2.varValue:.2f} toneladas")
print(f"Ganancia máxima =  ${maximizar.objective.value():.2f}")
print()
print(" = " *45)


# Verificación
# Calculamos cuánto material de cada compuesto se gastó realmente con la producción óptima
comp_A = 3*x1.varValue + 4*x2.varValue
comp_B = 2*x1.varValue + x2.varValue
comp_C = 5*x1.varValue + 3*x2.varValue

print()
print(f"Compuesto A usado: {comp_A:6.2f} kg     (Límite: 120 kg)")
print(f"Compuesto B usado: {comp_B:6.2f} kg     (Límite: 80 kg)")
print(f"Compuesto C usado: {comp_C:6.2f} kg     (Límite: 150 kg)")
```
