```python
# EJERCICIO 2: PRODUCCIÓN SEMANAL DE UNA FÁBRICA
# GINNETH SILVANA MARTÍNEZ URIBE

import numpy as np

# S_n+1 = 0.6*S_n + 0.2*M_n + 40
# M_n+1 = 0.1*S_n + 0.5*M_n + 20

def produccion_siguiente(S_actual, M_actual):
    S_sig = 0.6 * S_actual + 0.2 * M_actual + 40
    M_sig = 0.1 * S_actual + 0.5 * M_actual + 20
    return S_sig, M_sig

# Cantidades iniciales
S0 = 200
M0 = 80
n_simular = 10  # Número de semanas a calcular en la simulación

S = [S0]
M = [M0]

# Simulación de la producción
for i in range(n_simular):
    # Calculamos la siguiente semana usando la función
    S_sig, M_sig = produccion_siguiente(S[i], M[i])

    S.append(S_sig)
    M.append(M_sig)

# Resultado
print (" = " *45)
print ("EVOLUCIÓN DE PRODUCCIÓN SEMANAL")
print ("Estado Simulación finalizada")
print ()


# Mostrar el historial semana a semana
for i in range(len(S)):
    print (f"Semana {i}  Sillas = {S[i]:6.2f}    Mesas = {M[i]:6.2f}")
print (" = " *45)

# Verificación (Cálculo del equilibrio teórico)
A = np.array([[0.4, -0.2],
              [-0.1,  0.5]])
B = np.array([40, 20])

equilibrio = np.linalg.solve(A, B)
print()
print(f"Equilibrio teórico: Sillas = {equilibrio[0]:.2f}    Mesas = {equilibrio[1]:.2f}")
print(f"Valores semana {n_simular}:  Sillas = {S[-1]:.2f}    Mesas = {M[-1]:.2f}")



```
